## Irreversible Blur Overview
The `irreversible_blur` method has *three components*:

+ [Gaussian blur](https://en.wikipedia.org/wiki/Gaussian_blur)
+ [Pixelation](https://en.wikipedia.org/wiki/Pixelation)
+ [Random Noise](https://en.wikipedia.org/wiki/Image_noise)

These *3 components* will be applied to the image and can be adjusted as shown below (or completely disabled):

```python
# get libs
from blog_utils.images.transform import irreversible_blur
from blog_utils.images.display import display_images_in_grid

# get first cat
cat_n1 = cat_raw_images[0]

# blur
cat_n1_gaussian = irreversible_blur(
    cat_n1,
    **{
        "downscale_factor": 1,
        "kernel_size": 81,
        "noise_intensity": 0,
    }
)

# pixelate
cat_n1_pixelated = irreversible_blur(
    cat_n1,
    **{
        "downscale_factor": 0.08,
        "kernel_size": 1,
        "noise_intensity": 0,
    }
)

# noise
cat_n1_noise = irreversible_blur(
    cat_n1,
    **{
        "downscale_factor": 1,
        "kernel_size": 1,
        "noise_intensity": 200,
    }
)

# display cats
display_images_in_grid(
    [cat_n1, cat_n1_gaussian, cat_n1_pixelated, cat_n1_noise],
    ["Original", "Gaussian", "Pixelated", "Noise"],
    (300, 300),
    2,
    "Figure 1. Different Irreversible Blur Components on Example Cat Image.",
    bottom_title_coords=(0.5, -0.1),
    ax_title_y_coord=-0.2,
)
```

It goes without saying that you can adjust the different components to your target *blurred* threshold. Here is what the *example cat image* looks like with all 3 components combined:

```python
# combined blur components
cat_n1_combined = irreversible_blur(
    cat_n1,
    **{
        "downscale_factor": 0.08,
        "kernel_size": 81,
        "noise_intensity": 200,
    }
)

# display cats
display_images_in_grid(
    [cat_n1, cat_n1_combined],
    ["Original", "Combined"],
    (300, 300),
    1,
    "Figure 2. Combined Components for Irreversible Blur on Example Cat Image.",
    bottom_title_coords=(0.5, 0),
    ax_title_y_coord=-0.2,
)
```
